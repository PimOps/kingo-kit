# Welcome to SKKU Kingo Kit

This notebook verifies the shared PostgreSQL warehouse and introduces the two sample datasets. Credentials are already supplied to this container through environment variables.

```python
import os
import pandas as pd
from sqlalchemy import create_engine, text

engine = create_engine(os.environ["DATABASE_URL"])
with engine.connect() as connection:
    print(connection.execute(text("select version()")).scalar())
```

```python
query = """
select table_schema, count(*) as tables
from information_schema.tables
where table_type = 'BASE TABLE'
  and table_schema not in ('pg_catalog', 'information_schema', 'kingo_meta')
group by table_schema
order by table_schema
"""
pd.read_sql(query, engine)
```
